```python
X = df.drop(
    columns=['type', 'quality']
)

y = df['quality']
```

```python
y.value_counts().sort_index()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(
    k_neighbors=2,
    random_state=42
)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train,
    y_train
)
```

```python
print("Before SMOTE:")
print(y_train.value_counts().sort_index())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts().sort_index())
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

def evaluate_model(model, X_train, y_train, X_test, y_test):
    
    # Train
    model.fit(X_train, y_train)
    
    # Predict
    y_pred = model.predict(X_test)
    
    # Accuracy
    accuracy = accuracy_score(y_test, y_pred)
    
    print("Accuracy:", round(accuracy * 100, 2), "%")
    
    # Classification report
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
    
    return model, y_pred, accuracy
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_smote)
X_test_scaled = scaler.transform(X_test)
```

```python
logistic_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

logistic_model, logistic_pred, logistic_accuracy = evaluate_model(
    logistic_model,
    X_train_scaled,
    y_train_smote,
    X_test_scaled,
    y_test
)
```

```python
from sklearn.tree import DecisionTreeClassifier

decision_tree = DecisionTreeClassifier(
    random_state=42
)

decision_tree, dt_pred, dt_accuracy = evaluate_model(
    decision_tree,
    X_train_smote,
    y_train_smote,
    X_test,
    y_test
)
```

```python
from sklearn.ensemble import RandomForestClassifier

random_forest = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

random_forest, rf_pred, rf_accuracy = evaluate_model(
    random_forest,
    X_train_smote,
    y_train_smote,
    X_test,
    y_test
)
```

```python
from sklearn.ensemble import ExtraTreesClassifier

extra_trees = ExtraTreesClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

extra_trees, et_pred, et_accuracy = evaluate_model(
    extra_trees,
    X_train_smote,
    y_train_smote,
    X_test,
    y_test
)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)

print("Original classes:")
print(encoder.classes_)

print("\nEncoded classes:")
print(np.unique(y_encoded))
```

```python
X_train_xgb, X_test_xgb, y_train_xgb, y_test_xgb = train_test_split(
    X,
    y_encoded,
    test_size=0.25,
    random_state=42,
    stratify=y_encoded
)
```

```python
smote_xgb = SMOTE(
    k_neighbors=2,
    random_state=42
)

X_train_xgb_smote, y_train_xgb_smote = smote_xgb.fit_resample(
    X_train_xgb,
    y_train_xgb
)
```

```python
import xgboost as xgb

xgb_model = xgb.XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    random_state=42,
    eval_metric='mlogloss'
)

xgb_model, xgb_pred, xgb_accuracy = evaluate_model(
    xgb_model,
    X_train_xgb_smote,
    y_train_xgb_smote,
    X_test_xgb,
    y_test_xgb
)
```

```python
xgb_pred_quality = encoder.inverse_transform(xgb_pred)

xgb_pred_quality[:10]
```

```python
import lightgbm as lgb

lgb_model = lgb.LGBMClassifier(
    n_estimators=300,
    learning_rate=0.05,
    random_state=42,
    verbosity=-1
)

lgb_model, lgb_pred, lgb_accuracy = evaluate_model(
    lgb_model,
    X_train_smote,
    y_train_smote,
    X_test,
    y_test
)
```

```python
results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Decision Tree',
        'Random Forest',
        'Extra Trees',
        'XGBoost',
        'LightGBM'
    ],
    
    'Accuracy': [
        logistic_accuracy,
        dt_accuracy,
        rf_accuracy,
        et_accuracy,
        xgb_accuracy,
        lgb_accuracy
    ]
})

results['Accuracy'] = results['Accuracy'] * 100

results = results.sort_values(
    by='Accuracy',
    ascending=False
)

results
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=results,
    x='Accuracy',
    y='Model'
)

plt.title("Model Accuracy Comparison")
plt.xlabel("Accuracy (%)")
plt.ylabel("Model")

plt.xlim(0, 100)

plt.show()
```
